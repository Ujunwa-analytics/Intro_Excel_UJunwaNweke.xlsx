# Grouping by aggregation

```python
import pandas as pd
```

```python
df = pd.read_excel(r"C:\Users\Heisgolden\Downloads\world_population_excel_workbook.xlsx")
```

```python
df.groupby('Continent')
```

```python
group_by_frame = df.groupby('Continent')
```

```python
group_by_frame.mean(numeric_only=True)
```

```python
df.groupby('Continent').count()
```

```python
group_by_frame.min()
```

```python
group_by_frame.max()
```

```python
group_by_frame.sum(numeric_only=True)
```

```python
df.groupby('Continent').agg({
    'Growth Rate': ['mean', 'max', 'min']})
```

```python
df.groupby('Continent').agg({
    '2022 Population': ['mean', 'max', 'sum'],
    'Density (per kmÂ²)': ['mean', 'max'],
    'Growth Rate': ['mean', 'min', 'max']
})
```
